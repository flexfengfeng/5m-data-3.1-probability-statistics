## Scenario: Is Your Ad's Click Rate Reliable?

You work at a digital marketing agency. A client shows you their data:

> "We showed our new banner ad to **200 people** — 60 clicked it. Is 30% a reliable click rate?"

Before you can answer, you need to understand the **Law of Large Numbers**.
Whether 200 people is "enough" depends on a fundamental principle of probability.

**By the end of this section you'll be able to:**
- Explain why more data gives more reliable probability estimates
- Calculate and compare mean, median, and mode
- Connect the Law of Large Numbers to real A/B testing decisions

```python
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore", category=FutureWarning)
np.random.seed(42)
sns.set_style('whitegrid')
print("✅ Libraries loaded — you're ready to go!")
```

## 🎯 The Law of Large Numbers

**The idea in plain English:**
> The more times you repeat an experiment, the closer your observed result gets to the true probability.

**In our scenario:** If the true click-through rate for banner ads in this industry is 25%,
you might see 30% from 200 people — just by chance. But show it to 20,000 people,
and you'll be much closer to the real number.

Let's simulate this with a coin flip first (where we *know* the true probability is 50%),
then connect it back to ad clicks.

### ⏸️ Pause and Predict

We're about to run experiments of increasing size — flipping a coin 2, 4, 8, 16... up to 4,096
times — and record what proportion came up heads each time.

**Before running the cell below, predict:**
- Which experiment size will be furthest from 50%? (small or large?)
- Will the proportion always get closer to 50% as n increases, or can it jump around?

*Write your prediction here (double-click this cell to edit):*

```python
# Sample sizes — doubling each time from 2 to 4096 flips
ns = np.array([2, 4, 8, 16, 32, 64, 128, 256, 512, 1024, 2048, 4096])

# For each sample size: flip n coins, count how many are heads
heads_count = [np.random.binomial(n, 0.5) for n in ns]

# Convert counts to proportions (what fraction came up heads?)
proportion_heads = heads_count / ns

# Show the raw numbers
for n, p in zip(ns, proportion_heads):
    flag = '  ← far from 50%' if abs(p - 0.5) > 0.15 else ''
    print(f"  {n:5d} flips → {p:.3f} heads{flag}")
```

```python
# Plot: how proportion converges to 0.5 as sample size grows
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(ns, proportion_heads, marker='o', color='steelblue', linewidth=2, label='Observed proportion')
ax.axhline(0.5, color='orange', linewidth=2, linestyle='--', label='True probability (50%)')
ax.set_xscale('log')
ax.set_xlabel('Number of coin flips (log scale)', fontsize=12)
ax.set_ylabel('Proportion of heads', fontsize=12)
ax.set_title('Law of Large Numbers — proportion converges to true probability', fontsize=13)
ax.legend()
ax.set_ylim(0, 1)
plt.tight_layout()
plt.show()
```

### 💡 What do you notice?

- **Small samples** (2, 4, 8 flips) bounce around wildly — sometimes 0%, sometimes 100%
- **Large samples** (1,024+ flips) stay very close to the true 50%
- The orange dashed line is the **true probability** — small samples can miss it badly

**Back to our scenario:**

The client's 200-person ad test might show 30% when the true rate is 25% —
not because the ad is better, but because 200 people isn't a big enough sample.

> **Rule of thumb in A/B testing:** Most platforms recommend at least **1,000 people per variant**
> before trusting a result.

## 📊 Measures of Central Tendency

Now let's look at three ways to describe the "centre" of a dataset.
Each tells you something slightly different.

**Our new scenario:** We ran 1,000 experiments, each showing the ad to 5 different people.
How many people clicked in each experiment?

```python
# Simulate 1,000 experiments: each shows the ad to 5 people
# Assume true click probability = 0.5 (for clean numbers)
n_experiments = 1000
clicks_per_experiment = np.random.binomial(5, 0.5, n_experiments)

# Preview the first 20 results
print("First 20 experiment results (clicks out of 5 people):")
print(clicks_per_experiment[:20])
print(f"\nTotal experiments: {len(clicks_per_experiment)}")
```

```python
# Visualise the distribution of results
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(clicks_per_experiment, bins=range(0, 8), color='steelblue',
        edgecolor='white', alpha=0.85, align='left')
ax.set_xlabel('Number of clicks (out of 5 people)', fontsize=12)
ax.set_ylabel('How often this happened', fontsize=12)
ax.set_title('Distribution of clicks across 1,000 experiments', fontsize=13)
ax.set_xticks(range(7))
plt.tight_layout()
plt.show()
```

### Mean — the average

The **mean** is the sum of all values divided by the count.
It's what most people mean by "average".

**Formula:** $\bar{x} = \frac{\sum x_i}{n}$

**When to use it:** When your data doesn't have extreme outliers.
It's the most common and most mathematically useful measure of centre.

```python
# Calculate mean: total clicks divided by number of experiments
mean_clicks = np.mean(clicks_per_experiment)

print(f"Mean clicks per experiment: {mean_clicks:.3f}")
print(f"Theoretical expectation:    2.500  (5 people × 50% = 2.5)")
print(f"Difference from theory:     {abs(mean_clicks - 2.5):.3f}")
```

### Median — the middle value

The **median** is the middle value when all data is sorted.
Half the values are above it, half are below.

**When to use it:** When you have outliers that would distort the mean.
For example, house prices are often reported as *median* because a handful of
very expensive properties would pull the mean unrealistically high.

```python
# Sort and find the middle value
sorted_clicks = np.sort(clicks_per_experiment)

print(f"Median clicks per experiment: {np.median(sorted_clicks)}")
print(f"Mean clicks per experiment:   {np.mean(sorted_clicks):.3f}")
print()
print("For symmetric distributions, mean and median are very close.")
print("They diverge when data is skewed (e.g. income, house prices).")
```

### Mode — the most common value

The **mode** is the value that appears most often.

**When to use it:** Categorical data, or when you want to know the most typical outcome.
"What's the most common number of clicks in a 5-person experiment?" 

```python
# Find the mode
mode_result = st.mode(clicks_per_experiment)
mode_value = mode_result[0]
mode_count = mode_result[1]

print(f"Mode: {mode_value} clicks")
print(f"This happened {mode_count} times out of {n_experiments} experiments")
print(f"({100 * mode_count / n_experiments:.1f}% of experiments)")
```

```python
# Visualise all three measures together
fig, ax = plt.subplots(figsize=(9, 4))
ax.hist(clicks_per_experiment, bins=range(0, 8), color='steelblue',
        edgecolor='white', alpha=0.7, align='left')
ax.axvline(np.mean(clicks_per_experiment), color='orange', linewidth=2.5,
           label=f'Mean = {np.mean(clicks_per_experiment):.2f}')
ax.axvline(np.median(clicks_per_experiment), color='green', linewidth=2.5,
           linestyle='--', label=f'Median = {np.median(clicks_per_experiment):.0f}')
ax.axvline(float(st.mode(clicks_per_experiment)[0]), color='red', linewidth=2.5,
           linestyle=':', label=f'Mode = {int(st.mode(clicks_per_experiment)[0])}')
ax.set_xlabel('Clicks per experiment', fontsize=12)
ax.set_ylabel('Frequency', fontsize=12)
ax.set_title('Mean, Median and Mode on the same distribution', fontsize=13)
ax.legend(fontsize=11)
ax.set_xticks(range(7))
plt.tight_layout()
plt.show()
```

### 💡 When do they differ?

For this symmetric distribution, mean ≈ median ≈ mode.
They only diverge when data is **skewed**.

Run the cell below to see what happens with realistic salary data,
where a small number of executives earn much more than everyone else:

```python
# Real-world example: monthly salaries (SGD) — skewed because of high earners
np.random.seed(99)
salaries = np.concatenate([
    np.random.normal(4500, 800, 900),    # Most employees: ~$4,500
    np.random.normal(12000, 2000, 80),   # Senior staff: ~$12,000
    np.random.normal(45000, 5000, 20),   # Executives: ~$45,000
]).clip(2000, 80000)

print(f"Mean salary:   ${np.mean(salaries):,.0f}   ← pulled up by executives")
print(f"Median salary: ${np.median(salaries):,.0f}   ← better reflects the typical employee")

fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(salaries, bins=50, color='steelblue', edgecolor='white', alpha=0.75)
ax.axvline(np.mean(salaries), color='orange', linewidth=2.5,
           label=f'Mean ${np.mean(salaries):,.0f}')
ax.axvline(np.median(salaries), color='green', linewidth=2.5, linestyle='--',
           label=f'Median ${np.median(salaries):,.0f}')
ax.set_xlabel('Monthly Salary (SGD)', fontsize=12)
ax.set_title('Skewed distribution: mean vs median', fontsize=13)
ax.legend(fontsize=11)
plt.tight_layout()
plt.show()
```

## ✅ Section 1 Summary

| Concept | What it means | Real-world use |
|---|---|---|
| **Law of Large Numbers** | More data → closer to true probability | Need 1,000+ users for a reliable A/B test |
| **Mean** | Sum ÷ count | Default "average" — use when no extreme outliers |
| **Median** | Middle value | House prices, salaries — robust to outliers |
| **Mode** | Most common value | Categorical data, most typical outcome |

**Key insight for our ad scenario:**
> 200 people is NOT enough to be confident about a 30% click rate.
> The Law of Large Numbers tells us we need far more data for the result
> to stabilise around the true value.
