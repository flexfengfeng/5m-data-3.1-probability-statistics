## 🟢 Tier 1: Guided — Customer Spend Analysis

**Scenario:** You work at TechFlow, a B2B software startup. You have monthly spend data for two customer groups:
- **Group A** — customers who converted from a free trial
- **Group B** — customers who purchased directly without a trial

The sales team recently spent **$50,000** on a campaign to attract and convert free-trial users. Your manager wants to know: do direct purchasers spend significantly more than trial converts? If the spend difference is large enough, it may justify continuing the campaign — but if it isn't, that $50,000 could be better spent elsewhere.

**Your job:** Before running each cell, read the comments and predict what the output will look like. After running, check your prediction.

````python
# --- Generate synthetic customer spend data ---
# Group A: free-trial converts — lower average spend, wider variation
group_a = np.random.normal(loc=85, scale=20, size=120)
# Group B: direct purchasers — higher average spend
group_b = np.random.normal(loc=105, scale=18, size=120)

# Clip to realistic values (no negative spend)
group_a = np.clip(group_a, 10, None)
group_b = np.clip(group_b, 10, None)

# Summary statistics for both groups
for label, data in [('Group A (Free Trial)', group_a), ('Group B (Direct)', group_b)]:
    print(f'{label}')
    print(f'  Mean:    ${np.mean(data):.2f}')
    print(f'  Median:  ${np.median(data):.2f}')
    print(f'  Std Dev: ${np.std(data):.2f}')
    print()
````

````python
# --- Visualise both distributions side by side ---
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: overlapping KDE to compare shapes
sns.kdeplot(group_a, ax=axes[0], label='Group A (Free Trial)', fill=True, alpha=0.4, color='steelblue')
sns.kdeplot(group_b, ax=axes[0], label='Group B (Direct)', fill=True, alpha=0.4, color='coral')
axes[0].set_title('Spend Distribution by Customer Group')
axes[0].set_xlabel('Monthly Spend ($)')
axes[0].legend()

# Right: box plot to compare medians and spread
axes[1].boxplot([group_a, group_b], labels=['Group A
(Free Trial)', 'Group B
(Direct)'],
                patch_artist=True,
                boxprops=dict(facecolor='lightyellow'),
                medianprops=dict(color='coral', linewidth=2))
axes[1].set_title('Spend Comparison — Box Plot')
axes[1].set_ylabel('Monthly Spend ($)')

plt.tight_layout()
plt.show()
````

````python
# --- Hypothesis test: is the difference real or random? ---
#
# H₀ (null hypothesis):      The two groups spend the same on average
# H₁ (alternative):          Group B spends more than Group A
#
# We use an independent samples t-test:
# - Two separate groups (independent samples)
# - Comparing their means
# - Data is roughly normally distributed

t_stat, p_value = stats.ttest_ind(group_a, group_b)

print(f'T-statistic: {t_stat:.3f}')
print(f'P-value:     {p_value:.6f}')
print()

alpha = 0.05
if p_value < alpha:
    print(f'✅ p = {p_value:.4f} < {alpha} — statistically significant.')
    print('We reject H₀. Group B spends significantly more than Group A.')
else:
    print(f'❌ p = {p_value:.4f} ≥ {alpha} — not statistically significant.')
    print('We cannot conclude the groups spend differently.')
````

## Tier 3 Solution — Website Traffic Anomaly Detection

````python
# --- Tier 3 complete solution ---
np.random.seed(99)
traffic = np.random.normal(loc=45000, scale=5000, size=90).astype(int)
traffic[22] = 12000
traffic[57] = 18000
traffic[81] = 11000
days = np.arange(1, 91)

# Step 1: Calculate z-scores
mean_t = np.mean(traffic)
std_t  = np.std(traffic)
z_scores = (traffic - mean_t) / std_t

# Step 2: Identify anomalies
threshold = -2
print('Flagged anomalous days (z-score < -2):')
print(f'{"Day":>5}  {"Visitors":>12}  {"Z-score":>8}')
print('-' * 30)
for i, (day, t, z) in enumerate(zip(days, traffic, z_scores)):
    if z < threshold:
        print(f'{day:>5}  {t:>12,}  {z:>8.2f}')

# Step 3: Line chart with anomalies highlighted
normal_mask = z_scores >= threshold
anomaly_mask = z_scores < threshold

fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(days, traffic, color='steelblue', linewidth=1.5, alpha=0.8, label='Daily traffic')
ax.scatter(days[anomaly_mask], traffic[anomaly_mask],
           color='red', zorder=5, s=80, label='Flagged anomaly (z < -2)')
ax.axhline(mean_t + threshold * std_t, color='orange', linestyle='--', linewidth=1.5,
           label=f'Threshold (mean - 2×std = {mean_t + threshold*std_t:,.0f})')
ax.set_xlabel('Day')
ax.set_ylabel('Daily Visitors')
ax.set_title('Website Traffic — Anomaly Detection using Z-Scores')
ax.legend()
plt.tight_layout()
plt.show()
````

**Sample interpretation:**

> Three anomalous days were detected: days 23, 58, and 82, all with z-scores below −5 — far beyond the −2 threshold. A z-score of −2 is a reasonable starting point because it captures values that would occur by chance only about 2.3% of the time in a normal distribution, making them worth investigating without flagging too many false alarms. However, the threshold should be calibrated to the business context: in a high-stakes system you might lower it to −1.5 to catch more potential issues; in a noisy dataset you might raise it to −3 to reduce false positives. I would recommend the engineering team cross-reference these three days with their deployment and incident logs.

## Exercise 1 Solution — Appointment Booking Simulation

````python
np.random.seed(42)
slots = ['Morning', 'Afternoon', 'Evening']
bookings = np.random.choice(slots, size=10_000)

unique, counts = np.unique(bookings, return_counts=True)

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(unique, counts, color=['steelblue', 'coral', 'mediumseagreen'], edgecolor='white')
ax.axhline(10_000 / 3, color='orange', linestyle='--', linewidth=2, label='Expected (1/3 = 3,333)')
ax.set_title('Simulated Appointment Slot Distribution (n=10,000)')
ax.set_ylabel('Number of Bookings')
ax.legend()
plt.tight_layout()
plt.show()

print('Empirical probabilities:')
for slot, count in zip(unique, counts):
    print(f'  {slot}: {count} bookings → {count/10_000:.4f}  (theoretical: {1/3:.4f})')
````

**Sample interpretation:**

> After 10,000 simulated bookings, each slot receives close to — but not exactly — one third of appointments, with deviations of less than 1%. This is the Law of Large Numbers in action: random variation is present but small relative to the sample size. With 100,000 bookings, the noise would average out further and empirical probabilities would sit even closer to the theoretical 1/3, because each additional booking contributes less marginal randomness to the overall proportion.

## Exercise 2 Solution — Comparing Patient Wait Times

````python
np.random.seed(42)
group_a = np.random.normal(loc=12, scale=4, size=100)
group_b = np.random.normal(loc=15, scale=4, size=100)

# Overlapping histograms
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(group_a, bins=20, alpha=0.6, color='steelblue', edgecolor='white', label='Group A (App)')
ax.hist(group_b, bins=20, alpha=0.6, color='coral', edgecolor='white', label='Group B (Phone)')
ax.set_xlabel('Wait Time (minutes)')
ax.set_ylabel('Number of Patients')
ax.set_title('Wait Time Distribution: App vs Phone Booking')
ax.legend()
plt.tight_layout()
plt.show()

# T-test
t_stat, p_value = stats.ttest_ind(group_a, group_b)
print(f'T-statistic: {t_stat:.3f}  |  P-value: {p_value:.6f}')
print(f'Significant at α=0.05: {p_value < 0.05}')

# 95% Confidence intervals
def ci_95(sample):
    n, mean, se = len(sample), np.mean(sample), stats.sem(sample)
    margin = se * stats.t.ppf(0.975, df=n-1)
    return mean, mean - margin, mean + margin

mean_a, lo_a, hi_a = ci_95(group_a)
mean_b, lo_b, hi_b = ci_95(group_b)
print(f'\nGroup A — mean: {mean_a:.2f} min, 95% CI: [{lo_a:.2f}, {hi_a:.2f}]')
print(f'Group B — mean: {mean_b:.2f} min, 95% CI: [{lo_b:.2f}, {hi_b:.2f}]')

# CI visualisation
fig, ax = plt.subplots(figsize=(6, 3))
for i, (label, mean, lo, hi, color) in enumerate([
    ('Group A (App)',   mean_a, lo_a, hi_a, 'steelblue'),
    ('Group B (Phone)', mean_b, lo_b, hi_b, 'coral')]):
    ax.plot([lo, hi], [i, i], color=color, linewidth=5, solid_capstyle='round')
    ax.plot(mean, i, 'o', color='white', markersize=10,
            markeredgecolor=color, markeredgewidth=2)
    ax.text(hi + 0.1, i, f'{mean:.1f} min', va='center')
ax.set_yticks([0, 1])
ax.set_yticklabels(['Group A (App)', 'Group B (Phone)'])
ax.set_xlabel('Wait Time (minutes)')
ax.set_title('95% Confidence Intervals')
plt.tight_layout()
plt.show()
````

**Sample interpretation:**

> The t-test returns a p-value well below 0.05, confirming the 3-minute difference is statistically significant and very unlikely to be due to random variation. The 95% confidence intervals for the two groups do not overlap, which reinforces this: we can be highly confident that app users genuinely wait less. The data supports recommending the app-based booking system. That said, statistical significance does not automatically mean the difference is large enough to matter clinically — the team should also consider whether 3 minutes is meaningful to patients and whether the app is equally accessible to all patient groups.

## Exercise 3 Solution — Clinical Correlation Analysis

````python
from sklearn.datasets import load_diabetes

diabetes = load_diabetes()
bmi = diabetes.data[:, 2]
bp  = diabetes.data[:, 3]

# Scatter with regression line
slope, intercept, r, p, _ = stats.linregress(bmi, bp)
x_line = np.linspace(bmi.min(), bmi.max(), 100)

fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(bmi, bp, alpha=0.5, color='steelblue', edgecolors='white', s=40)
ax.plot(x_line, slope * x_line + intercept, color='coral', linewidth=2.5, label='Regression line')
ax.set_xlabel('BMI (normalised)')
ax.set_ylabel('Blood Pressure (normalised)')
ax.set_title('BMI vs Blood Pressure')
ax.legend()
plt.tight_layout()
plt.show()

# Pearson r and R²
r_val, p_val = stats.pearsonr(bmi, bp)
r_sq = r_val ** 2
print(f'Pearson r:  {r_val:.4f}')
print(f'P-value:    {p_val:.6f}')
print(f'R²:         {r_sq:.4f}  → BMI explains {r_sq*100:.1f}% of variation in blood pressure')
````

**Sample interpretation:**

> The Pearson correlation is approximately r = 0.39, indicating a moderate positive relationship: patients with higher BMI tend to have higher blood pressure. The p-value is well below 0.05, confirming this correlation is statistically significant. However, R² ≈ 0.15 means BMI alone explains only about 15% of the variation in blood pressure — the majority is driven by other factors such as age, medication, and lifestyle. The researcher is justified in proceeding with a full study, but the design should control for confounders to isolate BMI's specific contribution.
