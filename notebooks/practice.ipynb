## 🟢 Tier 1: Guided — Customer Spend Analysis

**Scenario:** You work at a subscription software company. You have monthly spend data for customers who converted from a free trial (Group A) vs customers who purchased directly (Group B). Your job is to describe this data and test whether the two groups spend significantly differently.

**Your job in Tier 1:** Before running each cell, read the code and the comments, then predict what the output will look like. After running, check your prediction.

````python
# --- Generate synthetic customer spend data ---
# Group A: free-trial converts — tend to spend less, wider variation
# Group B: direct purchasers — tend to spend more, slightly less variation
group_a = np.random.normal(loc=85, scale=20, size=120)   # mean=$85, std=$20, 120 customers
group_b = np.random.normal(loc=105, scale=18, size=120)  # mean=$105, std=$18, 120 customers

# Clip values to be realistic (no negative spend)
group_a = np.clip(group_a, 10, None)
group_b = np.clip(group_b, 10, None)

print('Group A (Free Trial Converts)')
print(f'  Mean:   ${np.mean(group_a):.2f}')
print(f'  Median: ${np.median(group_a):.2f}')
print(f'  Std Dev: ${np.std(group_a):.2f}')
print()
print('Group B (Direct Purchasers)')
print(f'  Mean:   ${np.mean(group_b):.2f}')
print(f'  Median: ${np.median(group_b):.2f}')
print(f'  Std Dev: ${np.std(group_b):.2f}')
````

````python
# --- Visualise the distributions side by side ---
# A KDE (Kernel Density Estimate) plot shows the shape of the distribution
# smoothly, without the bin-size sensitivity of a histogram

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left plot: overlapping KDE to compare shapes
sns.kdeplot(group_a, ax=axes[0], label='Group A (Free Trial)', fill=True, alpha=0.4, color='steelblue')
sns.kdeplot(group_b, ax=axes[0], label='Group B (Direct)', fill=True, alpha=0.4, color='coral')
axes[0].set_title('Spend Distribution by Customer Group')
axes[0].set_xlabel('Monthly Spend ($)')
axes[0].legend()

# Right plot: box plots to compare medians and spread
# Box plots show: median (middle line), IQR (box), range (whiskers), outliers (dots)
spend_df = pd.DataFrame({
    'Spend': np.concatenate([group_a, group_b]),
    'Group': ['Free Trial'] * len(group_a) + ['Direct'] * len(group_b)
})
sns.boxplot(data=spend_df, x='Group', y='Spend', ax=axes[1], palette=['steelblue', 'coral'])
axes[1].set_title('Spend Distribution — Box Plot')
axes[1].set_ylabel('Monthly Spend ($)')

plt.tight_layout()
plt.show()
````

````python
# --- Hypothesis Test: Is the difference real or random? ---
#
# Null hypothesis (H₀): The two groups spend the same on average
# Alternative hypothesis (H₁): Group B spends more than Group A
#
# We use an independent samples t-test because:
# - The two groups are separate (independent)
# - We're comparing their means
# - The data is roughly normally distributed

t_stat, p_value = stats.ttest_ind(group_a, group_b)

print(f'T-statistic: {t_stat:.3f}')
print(f'P-value:     {p_value:.6f}')
print()

# Interpret the result in plain English
alpha = 0.05  # Our significance threshold
if p_value < alpha:
    print(f'✅ Result is statistically significant (p={p_value:.4f} < {alpha})')
    print('   The difference in spending between the two groups is unlikely to be random.')
    print('   We reject the null hypothesis.')
else:
    print(f'❌ Result is NOT statistically significant (p={p_value:.4f} > {alpha})')
    print('   We cannot rule out that the difference is due to random variation.')
    print('   We fail to reject the null hypothesis.')
````
